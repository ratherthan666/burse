```python
import datetime as dt
from time import sleep

    import matplotlib.dates as md
import matplotlib.pyplot as plt

from burse.exchange import Exchange, Commodity
```

```python
e = Exchange()
```

```python
e.add_commodity(Commodity("dřevo", 10, 12))
e.add_commodity(Commodity("ocel", 14, 15))
```

```python
for _ in range(2000):
    e.random_price_movement_all()
    sleep(1)
```

```python
t1 = [dt.datetime.fromtimestamp(h["time"]) for h in e["dřevo"].history]
t2 = [dt.datetime.fromtimestamp(h["time"]) for h in e["ocel"].history]
sp1 = [h["sell"] for h in e["dřevo"].history]
sp2 = [h["sell"] for h in e["ocel"].history]

plt.subplots_adjust(bottom=0.2)
plt.xticks( rotation=25 )
ax=plt.gca()
xfmt = md.DateFormatter('%Y-%m-%d %H:%M:%S')
ax.xaxis.set_major_formatter(xfmt)
plt.plot(t1, sp1, label="dřevo")
plt.plot(t2, sp2, label="ocel")
plt.legend()
plt.show();
```
